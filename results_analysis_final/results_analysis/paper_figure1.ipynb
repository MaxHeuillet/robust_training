```python
from load_results import load_result_dataset
import pandas as pd
import numpy as np
pn1 = '/Users/maximeheuillet/Desktop/robust_training/results/full_fine_tuning50'
final_data = load_result_dataset(pn1)
final_data = [{**d, 'ft_strategy': 'FFT (50 epochs)'} for d in final_data]
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import math

import matplotlib as mpl
mpl.rcParams.update({
    'font.size': 13,
    'axes.labelsize': 13,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'legend.fontsize': 15,
    'legend.title_fontsize': 15,
})

pretty_color_labels = {
    'supervised': 'Sup.',
    'self-supervised': 'Self-sup.',
    'self-supervised (multimodal)': 'Multimodal',
    'supervised (robust)': 'Robust',
    'supervised (multistep)': 'Multistep',
    'fusion': 'Fusion'
}

# Pretty labels for model type (matches screenshot terminology)
pretty_model_type_labels = {
    'fully convolutional': 'Convolution',
    'fully attention'    : 'Attention',
    'hybrid'             : 'Hybrid'
}

# Pretty labels for model size (matches screenshot terminology)
pretty_size_labels = {
    'Tiny':  'Tiny',
    'Small': 'Small',
    'Base':  'Base'
}

# ── 1) Load ─────────────────────────────────────────────────────────────
data = pd.DataFrame(final_data)

# ── 2) Aggregate per configuration ──────────────────────────────────────
group_cols = ['backbone_name', 'loss_function', 'ft_strategy']
agg = (
    data.groupby(group_cols)
        .agg({
            'clean_acc': 'mean',
            'L1_acc': 'mean',
            'L2_acc': 'mean',
            'Linf_acc': 'mean',
            'common_acc': 'mean',
            'model_type': 'first',
            'model_size': 'first',
            'pre_training_strategy': 'first'
        })
        .reset_index()
)

# Normalise the strategy strings so they match the colour keys
agg['pre_training_strategy'] = (
    agg['pre_training_strategy']
        .astype(str).str.strip().str.lower()
)

# ── 3) Robust average ───────────────────────────────────────────────────
agg['robust_avg'] = agg[['L1_acc', 'L2_acc', 'Linf_acc','common_acc' ]].mean(axis=1)

# ── 4) Aesthetic mappings ───────────────────────────────────────────────
marker_map = {
    'fully convolutional': 'o',   # circle
    'fully attention'    : 'D',   # diamond
    'hybrid'             : '*'    # star
}

size_map = {'Tiny': 60, 'Small': 120, 'Base': 240}

# Colors matched to the screenshot
color_map = {
    'supervised'                    : '#1f77b4',  # blue
    'self-supervised'               : '#ff7f0e',  # orange
    'self-supervised (multimodal)'  : '#f7b6c2',  # pink
    'supervised (robust)'           : '#d62728',  # red
    'supervised (multistep)'        : '#9467bd',  # purple
    'fusion'                        : '#8c564b'   # brown
}

# TRADES is now bright green (matches screenshot)
edge_map = {
    'Classic AT': '#000000',   # solid black
    'TRADES'    : '#39FF14'    # bright neon green
}

# ── 5) Scatter plot ─────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 6))

for _, row in agg.iterrows():
    ax.scatter(
        row['clean_acc'],
        row['robust_avg'],
        marker   = marker_map.get(row['model_type'], 'o'),
        s        = size_map.get(row['model_size'], 80),
        color    = color_map.get(row['pre_training_strategy'], '#999999'),
        edgecolor= edge_map.get(row['loss_function'], '#333333'),
        linewidth=1.4,
        alpha    = 0.85
    )

ax.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)
ax.set_xlabel('Avg. clean accuracy')
ax.set_ylabel(r'Avg. robust accuracy' '\n' r'(common, $\ell_1$, $\ell_2$, $\ell_{\infty}$ perturbations)')

# Leave space on the right for the stacked legends
plt.tight_layout(rect=[0, 0, 0.78, 1])

# ── 6) Four non-overlapping legends ─────────────────────────────────────
# title_fontproperties makes the legend titles bold
legend_kw = dict(
    frameon=True,
    framealpha=0.9,
    borderpad=0.6,
    fontsize=13,
    title_fontproperties={'weight': 'bold', 'size': 14},
)

first  = ax.legend(
    [Line2D([0], [0], marker=m, color='w', markerfacecolor='#cccccc',
            markeredgecolor='#333333', markersize=10) for m in marker_map.values()],
    [pretty_model_type_labels[k] for k in marker_map.keys()],
    title='Arch. type',
    loc='upper left', bbox_to_anchor=(1.02, 1.02), **legend_kw)

second = ax.legend(
    [Line2D([0], [0], marker='o', color='w', markerfacecolor='#cccccc',
            markeredgecolor='#333333', markersize=math.sqrt(v))
     for v in size_map.values()],
    [pretty_size_labels[k] for k in size_map.keys()],
    title='Arch. size',
    loc='upper left', bbox_to_anchor=(1.02, 0.77), **legend_kw)

third  = ax.legend(
    [Line2D([0], [0], marker='o', color='w', markerfacecolor=c,
            markeredgecolor='#333333', markersize=10)
     for c in color_map.values()],
    [pretty_color_labels[k] for k in color_map.keys()],
    title='Pre-training',
    loc='upper left', bbox_to_anchor=(1.02, 0.52), **legend_kw)

fourth = ax.legend(
    [Line2D([0], [0], marker='o', color='w', markerfacecolor='#ffffff',
            markeredgecolor=c, markeredgewidth=2.0, markersize=10)
     for c in edge_map.values()],
    edge_map.keys(),
    title='Loss fn.',
    loc='upper left', bbox_to_anchor=(1.02, 0.11), **legend_kw)

ax.add_artist(first)
ax.add_artist(second)
ax.add_artist(third)

# ── 7) Save & show ──────────────────────────────────────────────────────
plot_path = './paper_figures/clean_vs_robust_scatter_enhanced_tight.png'
plt.savefig(plot_path, dpi=400, bbox_inches='tight', pad_inches=0.02)
plt.show()
```
